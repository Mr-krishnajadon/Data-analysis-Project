```python
def data_fetch_api():
    url='https://fakestoreapi.com/products'
    res=requests.get(url)
    res1=res.json()
    df=pd.DataFrame(res1)
    df['rate']=df['rating'].map(lambda x:x['rate'])
    df['count']=df['rating'].map(lambda x:x['count'])
    df=df[['id','title','price','category','rate','count']]
    return df
```

```python
df=data_fetch_api()
```
